# Modelling scattered light from in-field stars in WFI exposures

***

## What is this? 

One major contributor to the sky background level in astronomical images arises from stars that are inside the field of view of the detector. The light from point sources gets spread out following the Point Spread Function (PSF) of the instrument, generating characteristic spikes and extended halos of light. Often, the combined emission from multiple extended wings of the PSF generate a new layer of the sky background (`see Trujillo & Fliri 2016 <https://ui.adsabs.harvard.edu/abs/2016ApJ...823..123T/abstract>_`). 

There are several models of the Roman PSF available to generate the PSF layer. Currently (August 10th, 2026) ROSALIA implements the PSF model from `stpsf <https://stpsf.readthedocs.io/en/latest/>`_. However, in order to add support for brighter stars and increase computing efficiency, we are currently switching towards  `SUPERBACK <https://science.nasa.gov/mission/roman-space-telescope/superback-building-an-empirically-motivated-background-model-to-optimize-roman-observations/>`_ analytic PSF model.   

Additional information about ROSALIA can be found in the [ROSALIA for Roman documentation](https://rosalia.readthedocs.io/en/latest/) article on ReadTheDocs.

## Quick start

First we select a Roman mock image stored at the S3 bucket in Nexus. We generate a ROSALIA exposure object by using rosalia.core.exposure. ROSALIA will automatically detect that the path is a Nexus S3 file, but it also works with local fits or asdf files. 

```python
# With this line we generate the exposure object. This is the first step in any ROSALIA analysis. 
asdf_dir_uri = "s3://stpubdata/roman/nexus/soc_simulations/tutorial_data/roman-2026.2/"
input_name = asdf_dir_uri + 'r0003201001001001004_0004_wfi*_f106_cal.asdf' 

rosalia_exposure = rs.core.exposure(filename=input_name)
```

To estimate the PSF stray-light, ROSALIA automatically queries Gaia / 2MASS / WISE / Hipparcos and JPL / Horizons databases. Then it finds out which stars are inside each detector, and generates PSF models accoridng to its wavelength and position in the SCAs. The snapshots of the stars are stored in a temporary folder in the current directory. Then all the snapshots are reprojected to the original image frame, ready to be subtracted or masked. All this happens in the background after using the `rosalia_exposure.psf_background()` method. 

```python
# Here we execute the psf_background method to generate a model of all the stars in Roman WFI FOV. 
psf = rosalia_exposure.psf_background()
```
