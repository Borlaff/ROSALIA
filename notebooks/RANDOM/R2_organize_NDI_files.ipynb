```python
# Make the average of all the NDI files at 1 deg

deg1_ndi_old = glob.glob("/Users/aborlaff/NASA/ROSALIA/rosalia/CORE/NDI/RST/ndi_lvl1/lvl1*.fits")
deg1_ndi_fits_canvas = fits.open(deg1_ndi_old[0])

canvas = np.zeros(deg1_ndi_fits_canvas[0].data.shape)

for i in tqdm(range(len(deg1_ndi_old))):
    deg1_ndi_i = fits.open(deg1_ndi_old[i])
    canvas = canvas + deg1_ndi_i[0].data

canvas = canvas / len(deg1_ndi_old)
deg1_ndi_fits_canvas[0].data = canvas
deg1_ndi_fits_canvas.verify("silentfix")
deg1_ndi_fits_canvas.writeto("/Users/aborlaff/NASA/ROSALIA_DEPOT/1_deg_average_NDI.fits", overwrite=True)
```
