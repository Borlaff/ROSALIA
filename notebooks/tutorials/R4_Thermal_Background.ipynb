# Estimating thermal backgrounds with ROSALIA 

## What is this? 

The purpose of this notebook is to demonstrate the use of ROSALIA to retrieve the thermal backgrounds for a given exposure. 

## How to start

To run this notebook, please activate the "rosalia" kernel at the top right of your window.

## Imports

- *rosalia* for Roman/WFI stray-light analysis
- *matplotlib.pyplot* to plot the results. 

```python
import rosalia as rs
import matplotlib.pyplot as plt
```

```python
thermal_background = rs.thermal.get_thermal_background("F213", SCA=1)
plt.imshow(thermal_background, cmap="inferno", origin="lower")
plt.colorbar(label="Thermal Background (e-/s)")
```
