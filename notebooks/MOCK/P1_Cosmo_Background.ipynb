```python
# Generating a Mock Galaxy image for the Press Release mock image ROSALIA
# 
# Alejandro Borlaff - NASA Ames Research Center
# a.s.borlaff@nasa.gov - August 5, 2026

# Plan: 
# Make a mock exposure object. 
# Run stray-light model. 

import rosalia as rs 
from astropy.time import Time
import os
import pandas as pd
from astropy.io import fits
import matplotlib.pyplot as plt
os.chdir("/Users/aborlaff/NASA/ROSALIA/notebooks/MOCK")
import numpy as np
from astropy.io import fits
```

```python
r_name = "/Users/aborlaff/NASA/ROMAN/OLBERS/SIMS/DREAM/image_F184.fits"
g_name = "/Users/aborlaff/NASA/ROMAN/OLBERS/SIMS/DREAM/image_J129.fits"
b_name = "/Users/aborlaff/NASA/ROMAN/OLBERS/SIMS/DREAM/image_Z087.fits"

os.system("astwarp -K -h0 --scale=0.12,0.12 " + r_name) # Increasing the size of the background by 20% to enhance visibility 
os.system("astwarp -K -h0 --scale=0.12,0.12 " + g_name)
os.system("astwarp -K -h0 --scale=0.12,0.12 " + b_name)

```

```python
# Reference image 
ref_image = "/Users/aborlaff/NASA/ROSALIA/notebooks/MOCK/PR_mock_F087_Roman_RA_088.239_DEC_-48.533_MJD_61375.00000_PA_098.23_stray_drz_scaled.fits"

ref = fits.open(ref_image)
print(ref[0].data.shape)

r_scaled = fits.open(r_name.replace(".fits", "_scaled.fits"))
g_scaled = fits.open(g_name.replace(".fits", "_scaled.fits"))
b_scaled = fits.open(b_name.replace(".fits", "_scaled.fits"))

# r_back = 
import numpy as np


shape = ref[0].data.shape

# R band 
canvas = np.zeros(shape)
canvas = canvas + r_scaled[1].data[0:shape[0], 0:shape[1]]
canvas[np.isnan(ref[0].data)] = np.nan

canvas = -2.5*np.log10(canvas*1E-9*(0.12**2)/(0.101)**2) + 8.9 # To Jy/arcsec2
canvas_es = rs.detectors.mu2fe(mu=canvas, instrument="WFI", filter_name="F184", telescope="Roman")
rs.utils.save_fits(canvas_es.value, "/Users/aborlaff/NASA/ROSALIA/notebooks/MOCK/r_background.fits")

# G band 
canvas = np.zeros(shape)
canvas = canvas + g_scaled[1].data[0:shape[0], 0:shape[1]]
canvas[np.isnan(ref[0].data)] = np.nan
canvas = -2.5*np.log10(canvas*1E-9*(0.12**2)/(0.101)**2) + 8.9 # To Jy/arcsec2
canvas_es = rs.detectors.mu2fe(mu=canvas, instrument="WFI", filter_name="F129", telescope="Roman")
rs.utils.save_fits(canvas_es.value, "/Users/aborlaff/NASA/ROSALIA/notebooks/MOCK/g_background.fits")

# R band 
canvas = np.zeros(shape)
canvas = canvas + b_scaled[1].data[0:shape[0], 0:shape[1]]
canvas[np.isnan(ref[0].data)] = np.nan
canvas = -2.5*np.log10(canvas*1E-9*(0.12**2)/(0.101)**2) + 8.9 # To Jy/arcsec2
canvas_es = rs.detectors.mu2fe(mu=canvas, instrument="WFI", filter_name="F087", telescope="Roman")

rs.utils.save_fits(canvas_es.value, "/Users/aborlaff/NASA/ROSALIA/notebooks/MOCK/b_background.fits")
```
