# Estimating stray-light in Hubble Space Telescope

This is an example notebook demonstrating the use of ROSALIA to estimate stray-light from out-of-field sources. 

Note: ROSALIA's implementation for Hubble is very experimental. It depends on STRAYCOR GO program 17921, which is not yet completed. 
Due to the lack of detailed ray-tracing information about Hubble Space Telescope (Bely et al. 1991), the resolution of the stray-light 
baffling efficiency of the telescope is poor. These estimations are a placeholder for the completion of PID 17921 observation program using 
ACS/WFC in the environs of Jupiter. The analysis of those future observations will improve the present model.   

This effort was funded by STScI, HST Proposal. Cycle 30, ID. #17041
STRAYCOR: Stray-light background correction for HST ACS and WFC3 
https://ui.adsabs.harvard.edu/abs/2022hst..prop17041S/abstract


```python
import os 
import rosalia as rs
import matplotlib.pyplot as plt

plt.style.use(os.path.dirname(rs.__file__) + "/style/presi_style.mplstyle")
```

```python
filename = "/Users/aborlaff/Downloads/MAST_2026-05-08T2317/HST/j90za9qmq/j90za9qmq_flc.fits"
hst_exp = rs.core.exposure(filename=filename)
mos_data, mos_wcs = hst_exp.straylight()
```

```python
fig, ax = plt.subplots(figsize=(8,8))
im = ax.imshow(mos_data)
ax.set_title("Hubble ACS/WFC predicted stray-light")
ax.set_xlabel("X axis (pixel)")
ax.set_xlabel("Y axis (pixel)")
plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
```
